# Input padplane mapping

```python
new_maps = [ 
    'PadPlane/AutoPad-R1-RevA.sch',
    'PadPlane/AutoPad-R2-RevA-Pads.sch',
    'PadPlane/AutoPad-R3-RevA.sch',
]
old_maps = [ 
    'AutoPad-R1-RevA.sch',
    'AutoPad-R2-RevA-Pads.sch',
    'AutoPad-R3-RevA.sch',
]
```

```python
new_df_centroids, old_df_centroids = [], []
for imap in new_maps:
    new_df_centroids.append(pd.read_csv(f"{imap}.ChannelMapping.csv"))
for imap in old_maps:
    old_df_centroids.append(pd.read_csv(f"{imap}.ChannelMapping.csv"))
```

```python
import matplotlib.pyplot as plt

def analysis(data: pd.DataFrame):
    fig, ax = plt.subplots(3, 3, figsize=(15, 6))
    for i, feature in enumerate(['x', 'y', 'Radius']):
        x, y = divmod(i, 3)
        data[feature].plot.hist(ax=ax[x][y], ylabel='', title=feature, color='skyblue', logy=True)
```

```python
analysis(new_df_centroids[0])
```

```python
for i in range(3):
    new_df_centroids[i]['r2'] = new_df_centroids[i]['x']**2 + new_df_centroids[i]['y']**2
    new_df_centroids[i]['r'] = np.sqrt(new_df_centroids[i]['r2'])
```

```python
# plotting 3 2d plot for i in range(3) (for each module) of new_df_centroids[i]['r'] vs new_df_centroids[i]['Radius']

for i in range(3):
    plt.figure(figsize=(8, 6))
    plt.scatter(new_df_centroids[i]['Radius'], new_df_centroids[i]['PadR']/10, alpha=0.5)
    plt.xlabel('Radius')
    plt.ylabel('r [cm]')
    plt.title(f'2D plot of r vs Radius for Module {i}')
    plt.grid()
    plt.show()
```

```python
# plotting 3 2d plot for i in range(3) (for each module) of new_df_centroids[i]['r'] vs new_df_centroids[i]['Radius']

for i in range(3):
    plt.figure(figsize=(8, 6))
    plt.scatter(old_df_centroids[i]['Radius'], old_df_centroids[i]['PadR']/10, alpha=0.5)
    plt.xlabel('Radius')
    plt.ylabel('r [cm]')
    plt.title(f'2D plot of r vs Radius for Module {i}')
    plt.grid()
    plt.show()
```
